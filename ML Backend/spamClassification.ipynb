```python
df = df.drop_duplicates()
```

```python
df['charCount'] = df['Message'].str.len()
df['wordCount'] = df['Message'].str.split().str.len()
# df['sentenceCount'] = df['Message'].str.count(r'[.!?]') + 1
df["digitCount"] = df["Message"].str.count(r"\d")
df["urlCount"] = df["Message"].str.count(r"http|www")
df["currencyCount"] = df["Message"].str.count(r"£|\$|₹")
df["exclamationCount"] = df["Message"].str.count("!")
df.groupby("Category")[["charCount","wordCount","digitCount","urlCount","currencyCount"]].mean()
```

```python
spamOnly = df[df['Category']=='spam']
```

```python
spamOnly
```

```python
spamOnly.shape
```

```python
spamWordCount={}
for i in spamOnly.Message:
    # print(i)
    # print("\n")
    singleLine = i.split(' ')
    for j in singleLine:
        if j in spamWordCount:
            spamWordCount[j]=spamWordCount[j]+1
        else:
            spamWordCount[j]=1
    
print(spamWordCount)

# spamWordCount = spamOnly["Message"].str.split().explode().value_counts()
```

```python
sortedSpamWordCount = dict(sorted(spamWordCount.items(), key=lambda item: item[1],reverse=True))
```

```python
print(sortedSpamWordCount)
```
