```python
df['charCount'] = df['Message'].str.len()
df['wordCount'] = df['Message'].str.split().str.len()
# df['sentenceCount'] = df['Message'].str.count(r'[.!?]') + 1
df["digitCount"] = df["Message"].str.count(r"\d")
df["urlCount"] = df["Message"].str.count(r"http|www")
df["currencyCount"] = df["Message"].str.count(r"£|\$|₹")
df["exclamationCount"] = df["Message"].str.count("!")
df.groupby("Category")[["charCount","wordCount","digitCount","urlCount","currencyCount"]].mean()
```

```python
spamWordCount={}
for i in spamOnly.Message:
    # print(i)
    # print("\n")
    singleLine = i.split(' ')
    for j in singleLine:
        if j in spamWordCount:
            spamWordCount[j]=spamWordCount[j]+1
        else:
            spamWordCount[j]=1
    
print(spamWordCount)

# spamWordCount = spamOnly["Message"].str.split().explode().value_counts()
```

```python
from sklearn.model_selection import train_test_split

target = df['spam']
df.drop(columns=['spam'],inplace=True)

xTrain, xTest, yTrain, yTest = train_test_split(df,target,test_size=0.25,random_state=42,stratify=target)
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
from scipy.sparse import hstack
from sklearn.preprocessing import MinMaxScaler


tfV = TfidfVectorizer()
xTrainCount = tfV.fit_transform(xTrain['Message'])
extraFeatures = xTrain[
    ['charCount',
     'wordCount',
     'digitCount',
     'urlCount',
     'currencyCount',
     'exclamationCount']
]

# xTrainFinal = hstack([xTrainCount, extraFeatures])


xTestCount = tfV.transform(xTest["Message"])

extraTest = xTest[
    ['charCount',
     'wordCount',
     'digitCount',
     'urlCount',
     'currencyCount',
     'exclamationCount']
]

# xTestFinal = hstack([xTestCount, extraTest])



scaler = MinMaxScaler()

extraTrain = scaler.fit_transform(extraFeatures)
extraTest = scaler.transform(extraTest)


xTrainFinal = hstack([xTrainCount, extraTrain])
xTestFinal = hstack([xTestCount, extraTest])

model = MultinomialNB()

model.fit(xTrainFinal,yTrain)

pred = model.predict(xTestFinal)

accuracy_score(yTest, pred)
```

```python
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

print(classification_report(yTest, pred))
print(confusion_matrix(yTest, pred))
```

```python
countV = CountVectorizer()
xTrainCount = countV.fit_transform(xTrain['Message'])
extraFeatures = xTrain[
    ['charCount',
     'wordCount',
     'digitCount',
     'urlCount',
     'currencyCount',
     'exclamationCount']
]

# xTrainFinal = hstack([xTrainCount, extraFeatures])


xTestCount = countV.transform(xTest["Message"])

extraTest = xTest[
    ['charCount',
     'wordCount',
     'digitCount',
     'urlCount',
     'currencyCount',
     'exclamationCount']
]

# xTestFinal = hstack([xTestCount, extraTest])
scaler = MinMaxScaler()

extraTrain = scaler.fit_transform(extraFeatures)
extraTest = scaler.transform(extraTest)

xTrainFinal = hstack([xTrainCount, extraTrain])
xTestFinal = hstack([xTestCount, extraTest])

model = MultinomialNB()

model.fit(xTrainFinal,yTrain)
# model.fit(xTrainFinal,yTrain)

pred = model.predict(xTestFinal)

# accuracy_score(yTest, pred)

emails = [
    'Hey mohan, can we get together to watch footbal game tomorrow?',
    'Upto 20% discount on parking, exclusive offer just for you. Dont miss this reward!',
    'Get free million dollar prize in seconds sign up now !',
    'Games are waiting for you play games like playing in heaven !! dont miss this, see you soon',
    'Congratulations! You won a FREE iPhone. Click here to claim your prize now. Limited time offer.'
]
searchDf = pd.DataFrame(emails,columns=['Message'])
searchMessageCount = countV.transform(emails)

searchDf['charCount'] = searchDf['Message'].str.len()
searchDf['wordCount'] = searchDf['Message'].str.split().str.len()
searchDf["digitCount"] = searchDf['Message'].str.count(r"\d")
searchDf["urlCount"] = searchDf['Message'].str.count(r"http|www")
searchDf["currencyCount"] = searchDf['Message'].str.count(r"£|\$|₹")
searchDf["exclamationCount"] = searchDf['Message'].str.count("!")
extraSearchTest = searchDf[
    ['charCount',
     'wordCount',
     'digitCount',
     'urlCount',
     'currencyCount',
     'exclamationCount']
]
extraSearchTest = scaler.transform(extraSearchTest)
searchFinal = hstack([searchMessageCount, extraSearchTest])
model.predict(searchFinal)
```
